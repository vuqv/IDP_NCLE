```python
import pandas as pd
import numpy as np
import os
from pathlib import Path
import shutil
import matplotlib.pyplot as plt
```

```python
idrome = pd.read_excel('IDRome_DB_reanalyzed.xlsx')
df = pd.read_csv('Entanglement/order_calculated.csv', header=None, names=['Uniprot', 'Length'])
df
```

```python
def calculate_entanglement_probability(seq_name):
    source_file = Path(f"XTC2DCD/{seq_name}/traj_results.txt")
    #if source_file does exist
    if source_file.is_file():
        #work when file exist
        res_file = np.loadtxt(source_file) # col 5 is maxGc
        if res_file.shape[0] == 1000:
            maxGC = res_file[:,5]
            ent_prob = np.sum([1 if g >=0.6 else 0 for g in maxGC])/len(maxGC)
#             print(f"{seq_name} NOT FINISH")

            return ent_prob
        else:
            print(f"{seq_name} NOT FINISH")
            return None
    else:
        #this is when files does not exist
        print(f"{seq_name} NOT RUN")
        return None
    
```

```python
ent_prob_dict = {}
for seq in df.Uniprot:
    res = calculate_entanglement_probability(seq)
    ent_prob_dict[seq] = res
    print(f"{seq}: {res}")
```

```python
ent_prob_df = pd.DataFrame({'seq_name': ent_prob_dict.keys(), 'ent_prob': ent_prob_dict.values()})
# ent_prob_df
ent_valid = ent_prob_df[ent_prob_df['ent_prob'].notna()]
```

```python
# plot histogram
bin_width= 0.01
bins = np.arange(0, 1, bin_width)
x = bins[:-1] + bin_width/2
h, _ = np.histogram(ent_valid.ent_prob, bins=bins, density=True)

# Verify normalization
print("Normalized histogram:", h)
print("Sum of bin heights times bin width (area under curve):", np.sum(h * bin_width))
plt.bar(x, h, alpha=0.7, width=bin_width)
```

# Merge entanglement information to IDRome

```python
df_merged = pd.merge(idrome, ent_valid, on='seq_name', how='inner')
df_merged.to_pickle('IDRome_DB_reanalyzed_entanglement.pkl')
df_merged.to_excel('IDRome_DB_reanalyzed_entanglement.xlsx', index=False)
```

```python
# import matplotlib.pyplot as plt
# import numpy as np
# import pandas as pd

# matplotlib settings for publication-quality plots
base_fontsize = 14
plt.rc('font', **{'family':'sans-serif','sans-serif':['Arial']})

# Define x and y limits for both panels
xlim_panel_a = (0.05, 0.67)
ylim_panel_a = (-0.05, 1)
xlim_panel_b = (0.05, 0.67)
ylim_panel_b = (-0.05, 1)

# Filter data for Panel (a)
# control for N, exclude short segment
df_merged_N = df_merged[df_merged['N'] > 100].copy()

# Prepare data for Panel (b)
# test_df = df_merged[df_merged['N'] >= 100].copy()
bin_width = 0.05
bins = pd.interval_range(start=0, end=0.66, freq=bin_width, closed='left')
df_merged_N['nu_bin'] = pd.cut(df_merged_N['nu'], bins=bins)

# Calculate the mean and standard error of the mean (SEM) for each bin
grouped_data = df_merged_N.groupby('nu_bin').agg(
    mean_ent_prob=('ent_prob', 'mean'),
    sem_ent_prob=('ent_prob', 'sem')
).reset_index()

# Format bin labels and calculate bin midpoints
def format_bin_label(interval):
    left = f"{interval.left:.2f}".rstrip('0').rstrip('.')
    right = f"{interval.right:.2f}".rstrip('0').rstrip('.')
    return f"[{left}, {right})"

grouped_data['nu_bin'] = grouped_data['nu_bin'].apply(format_bin_label)
grouped_data['bin_midpoint'] = grouped_data['nu_bin'].apply(
    lambda x: (float(x.split(',')[0][1:]) + float(x.split(',')[1][:-1])) / 2
)

# Create the two-panel figure
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# Panel (a): Scatter plot of nu vs ent_prob
axes[0].scatter(df_merged_N['nu'], df_merged_N['ent_prob'], s=2, alpha=0.6)
axes[0].set_xlabel('$\\nu$', fontsize=base_fontsize)
axes[0].set_ylabel('$P_{entanglement}$', fontsize=base_fontsize)
axes[0].set_xlim(xlim_panel_a)
axes[0].set_ylim(ylim_panel_a)
axes[0].set_xticks(np.arange(0.1, 0.7, 0.1))
axes[0].tick_params(labelsize=base_fontsize, length=8, width=1.5)
axes[0].spines['top'].set_visible(False)
axes[0].spines['right'].set_visible(False)
axes[0].spines['left'].set_linewidth(2)
axes[0].spines['bottom'].set_linewidth(2)

# Panel (b): Mean and SEM plot for ent_prob by nu_bin
axes[1].errorbar(grouped_data['bin_midpoint'], grouped_data['mean_ent_prob'], 
                 yerr=grouped_data['sem_ent_prob'], fmt='o', capsize=3, alpha=0.7)
axes[1].set_xlabel('$\\nu$', fontsize=base_fontsize)
axes[1].set_ylabel('<$P_{entanglement}$>', fontsize=base_fontsize)
axes[1].set_xlim(xlim_panel_b)
axes[1].set_ylim(ylim_panel_b)
axes[1].set_xticks(grouped_data['bin_midpoint'][1:])
axes[1].set_xticklabels(grouped_data['nu_bin'][1:], rotation=45, ha='right')
axes[1].tick_params(axis='y', labelsize=base_fontsize, length=8, width=1.5)
axes[1].tick_params(axis='x', labelsize=base_fontsize - 2, length=8, width=1.5)
axes[1].spines['top'].set_visible(False)
axes[1].spines['right'].set_visible(False)
axes[1].spines['left'].set_linewidth(2)
axes[1].spines['bottom'].set_linewidth(2)

# Adjust layout for publication-quality formatting
plt.tight_layout()

# Save the figure as a high-quality image (optional)
# plt.savefig("two_panel_plot.png", dpi=300)

# Show the plot
plt.show()
```
