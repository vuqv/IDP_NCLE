```python
import pandas as pd
import pickle
```

```python
with open('go_terms.pkl', 'rb') as f:
    id_to_name = pickle.load(f)
    
# id_to_name
```

```python
df_cell = pd.read_pickle('cellcom_significant.pkl')
df_cell_table = df_cell[['GO', 'effect_size', 'adj_pvalue']]
GO_names = []
for _, row in df_cell_table.iterrows():
    name = id_to_name[row['GO']]
    GO_names.append(name)
    
df_cell_table['Name'] = GO_names
df_cell_table = df_cell_table[['GO', 'Name', 'effect_size', 'adj_pvalue']]
df_cell_table
```

```python
df_cell_table.to_excel('CellCom.xlsx')
```
